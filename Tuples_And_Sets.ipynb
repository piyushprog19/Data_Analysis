```python
#symbol for set{}
marks = {34, 43, 43, 43, 54, 65}
print(marks)
marks.add(True)
print(marks)
marks.update([12, 54, 87, 90])
print(marks)
```

```python
student = {"piyush", "Monkey D. Luffy", "Zoro"}
print(type(student))
print(student)
student.remove("Monkey D. Luffy")
print(student)
student.discard("Zoro")
print(student)
student.clear()
print(student)
```

```python
#Set opration
python_student = {"Amit", "Mandeep", "Randeep", "Jin", "Kazama"}
Java_student = {"Piyush", "Agarwal", "Senku", "Mandeep", "Amit"}
result = python_student.union(Java_student)#union operation
print(result)
result = python_student.intersection(Java_student)#intersection operation
print(result)
result = python_student.difference(Java_student)#difference operation
print(result)
result = Java_student.difference(python_student)#difference operation
print(result)
result = python_student.symmetric_difference(Java_student)#symmetric difference operation
print(result)
```
